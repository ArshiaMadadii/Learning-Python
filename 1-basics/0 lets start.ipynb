```python
def how_many(w , input_string):
    count = 0
    for s in input_string:
        if s == w: count += 1
    if count == 0 :
        return "we don't have it!"
    else:
        return count

how_many1 = how_many("a","arshia")
print(how_many1)
how_many2 = how_many("a","Arshia")
print(how_many2)
```

```python
donation = {
    'Arshia' : 500 ,
    'Narges' : 450 ,
    'Sina'   :  50 ,
    'Arsalan': 360 ,
    'Mahkam' :  860,
    'Mohsen' : 220 ,
    'Ali'    : 230 ,
}

def donation_analysis(dic):
    count = 0
    total = 0
    max_donation = 0
    for person,values in dic.items():
        total += values
        count += 1
        if max_donation <= values:
            max_person  = person
            max_donation = values
        else : continue

    avg_donation = total / count
    return avg_donation , total , max_person , max_donation
avg , total , max, max_donation = donation_analysis(donation)
print(f"total donation is : {total}\nmax person is : {max} and donated {max_donation}$\naverage is : {int(avg)}$")
```

# *args
## some time we don't know how much input we need;this time we use *args

```python
def zarb(*args):
    res_zarb = 1
    for i in args:
        res_zarb = i * res_zarb
        # or we can use this code => res_zarb *= i
    return res_zarb

zarb = zarb(2,2,4)
print(zarb)
```

```python
class book():
    def __init__(self,name, page):
        self.name = name
        self.pages = page #instance is pointing to self
        #page is attribute
    def open(self):
        print(f"opened the {self.name} on last page, last page is {self.pages}")
my_book = book(page=252,name="math") #my_book is instance
my_book.pages
my_book.open()
```

```python
class circle():
    pi = 3.1415926535897932384626433832795
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        area = self.radius * self.radius * circle.pi
        return area

c1 = circle(10)
print(f"area is {c1.area()}")
```
