# set

```python
my_set = set()
my_set.add(10)
my_set.add(115)
my_set.add(15)
my_set.add(10)
my_set.add("arshia")
print(f"type = {type(my_set)} and my set = {my_set}")
```

```python
math = set()
algebra = set()
math.add("arshia")
math.add("armin")
math.add("sina")
print(f"Math : {math}")
algebra.add("ali")
algebra.add("arshia")
algebra.add("arsalan")
print(f"Algebra : {algebra}")
```

```python
union_of_algebra_math =  algebra.union(math)
intersection_of_algebra_math = algebra.intersection(math)

print(f"Union of Algebra & Math : {union_of_algebra_math}")
print(f"Intersection of Algebra & Math : {intersection_of_algebra_math}")
```

```python
Persian_class =  set(['arshia','armin','arsalan','ali','darya','amir','erfan'])

print(f"Persian class : {Persian_class}")

union_of_algebra_Persian_class =  algebra.union(Persian_class)
intersection_of_algebra_Persian_class = algebra.intersection(Persian_class)

print(f"Union of Algebra & Persian : {union_of_algebra_Persian_class}")
print(f"Intersection of Algebra & Persian : {intersection_of_algebra_Persian_class}")
```

## FOR

```python
my_list = ['a',"b",123,6,'12']
num = 1
for this_item in my_list:
    print(f"number{num} : {this_item}")
    num = num + 1
num = 1
print('\n')
for this_item in my_list:
    print(f"2*number{num} : {this_item*2}")
    num = num + 1
```

```python
# for this code we use tuple
people = (('Arshia',23),('Arsalan',25),('Narges',22),('Mahkam',26),)
for person in people:
    name_person,age_person = person
    print(f"{name_person} has {age_person} old.")

# or we can write this code
print('\n')

for name_person1,age_person2 in people:
    print(f"{name_person1} has {age_person2} old.")

```
