```python
for num in range(1, 11): #show numbers from 1 to 10
    if num %2 == 0:
        print(f"this number '{num}' is even")
    else:
        print(f"this number '{num}' is odd")
```

```python
winiapp_group = (("Dr . naderi", "Manager"),
                  ("Arshia", "team leader"),
                  ("Arsalan", "team leader"),
                  ("Erfan", "team member"),
                  ("Mohsen", "intership student"),
                  ) #this is a tuple of tuples

for member in winiapp_group:
    print(member)
```
